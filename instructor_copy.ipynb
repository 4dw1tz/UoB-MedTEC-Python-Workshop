## Part 3: Calculations - Computing BMI (10 mins)

### 📖 Teaching Point:
Python can do math! We can calculate BMI using the formula:

**BMI = weight (kg) ÷ height² (m²)**

In Python: `**` means "to the power of" (so `height ** 2` means height squared)

```python
# BMI Calculation
weight_kg = 70.5
height_m = 1.65

# Calculate BMI
bmi = weight_kg / (height_m ** 2)

print("Weight:", weight_kg, "kg")
print("Height:", height_m, "m")
print("BMI:", bmi)
```

```python
# Let's make it look nicer with rounding!
# round(number, decimal_places) rounds to a specific number of decimals

weight_kg = 70.5
height_m = 1.65

bmi = weight_kg / (height_m ** 2)
bmi_rounded = round(bmi, 1)  # Round to 1 decimal place

print("Your BMI is:", bmi_rounded)
```

## Part 4: User Input - Interactive Calculator (5 mins)

### 📖 Teaching Point:
The `input()` function asks the user to type something. We use `float()` to convert text to a number with decimals.

**Note:** When students run this, they need to type in the box that appears and press Enter.

```python
# Interactive BMI Calculator!
print("=== BMI Calculator ===")

# Get input from user
weight_kg = float(input("Enter weight in kg: "))
height_m = float(input("Enter height in meters: "))

# Calculate BMI
bmi = round(weight_kg / (height_m ** 2), 1)

# Display result
print("\nYour BMI is:", bmi)
```

## Part 5: Making Decisions - Health Categories (15 mins)

### 📖 Teaching Point:
Computers can make decisions using `if`, `elif` (else if), and `else`. This is how we give health advice based on BMI!

**BMI Categories (WHO):**
- Below 18.5 → Underweight
- 18.5 to 24.9 → Normal weight
- 25.0 to 29.9 → Overweight
- 30.0 and above → Obese

**Important:** Show the indentation (spaces before the print). Python uses indentation to know what's inside the if block.

```python
# Simple example of if/else
bmi = 22.5

if bmi < 18.5:
    print("Underweight")
else:
    print("Not underweight")
```

```python
# Full BMI classification with elif
bmi = 27.3

if bmi < 18.5:
    print("Category: Underweight")
    print("Advice: Consider consulting a nutritionist.")
elif bmi < 25:
    print("Category: Normal weight")
    print("Advice: Great job! Maintain your healthy lifestyle.")
elif bmi < 30:
    print("Category: Overweight")
    print("Advice: Consider increasing physical activity.")
else:
    print("Category: Obese")
    print("Advice: Please consult your healthcare provider.")
```

## 🎉 Part 6: Complete BMI Calculator with Health Advice!

### 📖 Teaching Point:
Now let's put everything together! This is the complete application they've built.

**This is what they should feel proud of!**

```python
# ========================================
# 🏥 COMPLETE BMI CALCULATOR & ADVISOR 🏥
# ========================================

print("╔════════════════════════════════════╗")
print("║     BMI Calculator & Advisor       ║")
print("║        Medical Python 2026         ║")
print("╚════════════════════════════════════╝")
print()

# Get patient information
name = input("Enter patient name: ")
weight_kg = float(input("Enter weight in kg: "))
height_m = float(input("Enter height in meters: "))

# Calculate BMI
bmi = round(weight_kg / (height_m ** 2), 1)

# Display results
print("\n" + "="*40)
print("PATIENT REPORT")
print("="*40)
print(f"Patient: {name}")
print(f"Weight: {weight_kg} kg")
print(f"Height: {height_m} m")
print(f"BMI: {bmi}")
print("-"*40)

# Provide health category and advice
if bmi < 18.5:
    print("📊 Category: UNDERWEIGHT")
    print("💡 Advice: Consider consulting a nutritionist.")
    print("   Ensure adequate caloric intake and nutrition.")
elif bmi < 25:
    print("📊 Category: NORMAL WEIGHT ✅")
    print("💡 Advice: Excellent! Maintain your healthy lifestyle.")
    print("   Continue balanced diet and regular exercise.")
elif bmi < 30:
    print("📊 Category: OVERWEIGHT")
    print("💡 Advice: Consider lifestyle modifications.")
    print("   Increase physical activity, monitor diet.")
else:
    print("📊 Category: OBESE")
    print("💡 Advice: Please consult your healthcare provider.")
    print("   A comprehensive health assessment is recommended.")

print("="*40)
print("Thank you for using BMI Calculator!")
```

## 🚀 BONUS: Simple Patient Triage Tool (if time permits)

### 📖 Teaching Point:
This shows how Python could be used for clinical decision support. It's a simplified triage based on vital signs.

**Only show this if you have extra time or as a "look what else you could build!" teaser.**

```python
# ========================================
# 🚨 SIMPLE PATIENT TRIAGE TOOL 🚨
# ========================================

print("╔════════════════════════════════════╗")
print("║    Patient Triage Assessment       ║")
print("╚════════════════════════════════════╝")
print()

# Get vital signs
heart_rate = int(input("Enter heart rate (bpm): "))
temperature = float(input("Enter temperature (°C): "))
pain_level = int(input("Enter pain level (0-10): "))

# Assess urgency
print("\n" + "="*40)
print("TRIAGE ASSESSMENT")
print("="*40)

# Check for emergency conditions
if heart_rate > 120 or heart_rate < 50:
    print("🔴 URGENT: Abnormal heart rate detected!")
    print(f"   Heart rate: {heart_rate} bpm")
    
if temperature > 39.0:
    print("🔴 URGENT: High fever detected!")
    print(f"   Temperature: {temperature}°C")
elif temperature > 37.5:
    print("🟡 MODERATE: Elevated temperature")
    print(f"   Temperature: {temperature}°C")
else:
    print("🟢 Temperature normal")
    
if pain_level >= 8:
    print("🔴 URGENT: Severe pain reported!")
    print(f"   Pain level: {pain_level}/10")
elif pain_level >= 5:
    print("🟡 MODERATE: Significant pain")
    print(f"   Pain level: {pain_level}/10")
else:
    print("🟢 Pain level manageable")

print("="*40)
print("Assessment complete. Please review with clinician.")
```

## 🚀 BONUS: CHA₂DS₂-VASc Calculator (Challenge B Solution)

### 📖 Teaching Point:
This is a more advanced challenge that combines everything they've learned. It's a real clinical decision support tool used in cardiology.

**Only show if students finish early and want an extra challenge.**

```python
# ========================================
# 💊 CHA₂DS₂-VASc CALCULATOR - SOLUTION 💊
# ========================================

print("╔════════════════════════════════════════════╗")
print("║      CHA₂DS₂-VASc Score Calculator        ║")
print("║   Stroke Risk Assessment for AF Patients   ║")
print("╚════════════════════════════════════════════╝")
print()

# Initialise score
score = 0

# Get age and add points
age = int(input("How old is the patient? "))

if age >= 75:
    score = score + 2
elif age >= 65:
    score = score + 1

# Get gender
gender = input("Is the patient female? [y/n] ")

if gender == "y":
    score = score + 1

# Ask about risk factors
chf = input("Does the patient have congestive heart failure? [y/n] ")
if chf == "y":
    score = score + 1

htn = input("Does the patient have hypertension? [y/n] ")
if htn == "y":
    score = score + 1

diabetes = input("Does the patient have diabetes? [y/n] ")
if diabetes == "y":
    score = score + 1

vascular = input("Does the patient have vascular disease? [y/n] ")
if vascular == "y":
    score = score + 1

stroke_hx = input("Does the patient have history of stroke/TIA/thromboembolism? [y/n] ")
if stroke_hx == "y":
    score = score + 2

# Display results
print("\n" + "="*45)
print("RESULTS")
print("="*45)
print(f"CHA₂DS₂-VASc Score: {score}")
print("-"*45)

# Recommendation based on score and gender
if score == 0:
    print("💊 Recommendation: No anticoagulation required")
elif gender == "y":  # Female
    if score == 1:
        print("💊 Recommendation: No anticoagulation required")
    elif score == 2:
        print("💊 Recommendation: Consider anticoagulation")
    else:
        print("💊 Recommendation: Prescribe anticoagulation")
else:  # Male
    if score == 1:
        print("💊 Recommendation: Consider anticoagulation")
    else:
        print("💊 Recommendation: Prescribe anticoagulation")

print("="*45)
print("Please discuss with patient and review clinically.")
```
