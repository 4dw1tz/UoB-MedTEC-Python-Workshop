## Part 3: Python for Medical Data Analysis (5 mins)

### 📖 Teaching Point:
This is where we pivot from basic Python to data analysis. Explain that pandas is a library (a toolkit someone else wrote) that makes it easy to work with spreadsheet-like data.

**Key points to make:**
- pandas and matplotlib come **pre-installed** in Google Colab — no setup needed
- A DataFrame is just like an Excel spreadsheet with rows and columns
- `.head()` shows the first few rows (like peeking at the top of a spreadsheet)
- `.shape` tells you how many rows and columns there are

### 🙋 FAQ: "What's a library?"
> A library is a collection of pre-written code that someone else made. Instead of writing everything from scratch, we `import` their tools. Think of it like using a calculator app instead of doing long division by hand.

### 🙋 FAQ: "Do I need to install pandas?"
> No! Google Colab comes with pandas and matplotlib already installed. Just run the `import` cell.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

### Loading the Dataset

### 📖 Teaching Point:
Walk through the dataset creation. Explain that in real research you'd load a CSV file, but we're creating the data inline so there's nothing to download.

**Explain the data:** 100 patients in a kidney stone trial. Two treatments (A = open surgery, B = less invasive). We recorded stone size and whether it was successful.

### 🙋 FAQ: "Is this real data?"
> The numbers are inspired by a famous real study (Charig et al., 1986) that compared open surgery vs percutaneous nephrolithotomy for kidney stones. We've simplified it to 100 patients but the paradox is real.

```python
# Clinical trial dataset: 100 patients treated for kidney stones
patients = pd.DataFrame({
    "patient_id": range(1, 101),
    "treatment": ["A"]*50 + ["B"]*50,
    "stone_size": ["small"]*8 + ["large"]*42 + ["small"]*42 + ["large"]*8,
    "success": [1]*7 + [0]*1 + [1]*30 + [0]*12 + [1]*36 + [0]*6 + [1]*5 + [0]*3
})

# .head(10) shows the first 10 rows
patients.head(10)
```

### 📖 Quick Dataset Breakdown (for your reference):

| Group | Treatment | Stone Size | Count | Successes | Failures | Rate |
|-------|-----------|-----------|-------|-----------|----------|------|
| 1 | A | small | 8 | 7 | 1 | 87.5% |
| 2 | A | large | 42 | 30 | 12 | 71.4% |
| 3 | B | small | 42 | 36 | 6 | 85.7% |
| 4 | B | large | 8 | 5 | 3 | 62.5% |

**Overall:** A = 37/50 = 74%, B = 41/50 = 82%

The paradox: A is better in BOTH subgroups but worse overall because A got 84% hard cases (large stones) while B got 84% easy cases (small stones).

```python
# Exploring the data
print("Dataset shape:", patients.shape)
print()
print("Success rate overall:", round(patients["success"].mean() * 100, 1), "%")
print()
print("Patients per treatment:")
print(patients["treatment"].value_counts())
print()
print("Stone sizes:")
print(patients["stone_size"].value_counts())
```

### ✏️ Exercise 3: Explore the Data

**Instructor Solution:**

### 🙋 FAQ: "What's the difference between .sum() and .count()?"
> `.sum()` adds up all the values — since success is 1 or 0, summing gives us the number of successes. `.count()` counts all rows regardless of value. `.mean()` gives the proportion (same as sum/count).

```python
# SOLUTION: How many patients had a successful treatment?
print("Successful treatments:", patients["success"].sum())
print("Out of total:", patients["success"].count())
print("Success rate:", round(patients["success"].mean() * 100, 1), "%")
```

## Part 4: Filtering Data (5 mins)

### 📖 Teaching Point:
Filtering is one of the most powerful tools in pandas. The syntax `df[df["col"] == val]` can look confusing at first — break it down:

1. `patients["treatment"] == "A"` → creates a True/False list for every row
2. `patients[...]` → keeps only the True rows

**Analogy:** It's like asking a filing cabinet: "give me only the folders where the treatment was A"

### 🙋 FAQ: "Why is it == not =?"
> Single `=` is for **assigning** values (like `x = 5`). Double `==` is for **comparing** values (like "is x equal to 5?"). This trips up beginners — expect this question!

```python
# Filter: only patients who received Treatment A
treatment_a = patients[patients["treatment"] == "A"]
print("Treatment A patients:")
print(treatment_a.head())
print()

# Filter: only patients with small stones
small_stones = patients[patients["stone_size"] == "small"]
print("Small stone patients:")
print(small_stones.head())
```

### ✏️ Exercise 4: Filter the Data

**Instructor Solution:**

```python
# SOLUTION: Show only patients whose treatment failed
failed = patients[patients["success"] == 0]
print(f"Number of failed treatments: {len(failed)}")
print()
print(failed.head(10))
```

## Part 5: The Big Question — Which Treatment is Better? (5 mins)

### 📖 Teaching Point:
`.groupby()` is like a pivot table in Excel. It splits the data by a column, then calculates something for each group.

**Key moment:** After running this cell, ask students "So which treatment should we recommend?" Most will say B. Don't correct them yet — the twist comes in Part 7.

### 🙋 FAQ: "What does .mean() do here?"
> Since success is 1 or 0, the mean of those values is the proportion of successes. Multiply by 100 to get a percentage.

```python
# Compare overall success rates between treatments
success_by_treatment = patients.groupby("treatment")["success"].mean() * 100
success_by_treatment = round(success_by_treatment, 1)

print("=== SUCCESS RATES BY TREATMENT ===")
print()
print(success_by_treatment.to_string())
print()
print("🤔 So which treatment looks better?")
```

---

## Part 6: Visualisation (3 mins)

### 📖 Teaching Point:
Charts make the story instantly clear. matplotlib integrates with pandas — you can call `.plot()` directly on a Series or DataFrame.

**Reinforce the (intentionally misleading) narrative:** "Treatment B looks like the clear winner from this chart." This sets up the twist.

```python
# Bar chart: overall success rate by treatment
success_by_treatment.plot(kind="bar", color=["#2196F3", "#FF9800"], edgecolor="black")
plt.title("Overall Success Rate by Treatment")
plt.ylabel("Success Rate (%)")
plt.xlabel("Treatment")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print()
print("📊 Looks like Treatment B is the clear winner...")
print("   But should we trust this? Let's dig deeper.")
```

---

## Part 7: The Plot Twist 🔄 (8 mins)

### 📖 Teaching Point:
**THIS IS THE KEY MOMENT OF THE WORKSHOP.** Walk through each cell slowly.

1. First cell: Break down by treatment AND stone size → students should notice A is better in BOTH
2. Second cell: Visualise the grouped data to make it undeniable
3. Third cell: Reveal WHY — the uneven patient distribution

**Try to let students spot it themselves.** After showing the grouped numbers, ask: "Wait — what do you notice?" Give them 30 seconds before explaining.

### How to explain Simpson's Paradox simply:
> "Imagine two hospitals. Hospital A takes on the really sick patients, Hospital B only takes easy cases. Hospital B will have better survival stats — but that doesn't mean it's a better hospital. It just got easier patients. That's exactly what happened here with the kidney stone treatments."

### 🙋 FAQ: "How is this possible mathematically?"
> It's about **weighted averages**. Treatment A's overall rate is dragged down because 84% of its patients had large stones (harder cases, ~71% success). Treatment B's overall rate is pulled up because 84% of its patients had small stones (easier cases, ~86% success). The groups are weighted differently, so the averages flip.

### 🙋 FAQ: "Does this happen in real medicine?"
> Yes! The original kidney stone study is real. This also happens with:
> - Hospital mortality comparisons (hospitals taking sicker patients look worse)
> - Drug trial subgroup analyses
> - COVID-19 case fatality rates across countries with different age demographics
> - University admission bias (the famous Berkeley admissions case)

### 🙋 FAQ: "So how do you avoid being tricked by this?"
> Always look at **subgroup data** when comparing outcomes. In clinical trials, this is why we use **randomised controlled trials** (RCTs) — they ensure patient groups are balanced. And always check for **confounding variables** (in this case, stone size).

```python
# Success rates broken down by treatment AND stone size
detailed = patients.groupby(["treatment", "stone_size"])["success"].mean() * 100
detailed = round(detailed, 1)

print("=== SUCCESS RATES: TREATMENT × STONE SIZE ===")
print()
print(detailed.to_string())
print()
print("👀 Wait... look at those numbers carefully.")
print("   Treatment A is better for small stones AND large stones?!")
```

```python
# Let's visualise this properly
detailed_df = detailed.unstack()

detailed_df.plot(kind="bar", color=["#f44336", "#4CAF50"], edgecolor="black")
plt.title("Success Rate by Treatment AND Stone Size")
plt.ylabel("Success Rate (%)")
plt.xlabel("Treatment")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.legend(title="Stone Size")
plt.tight_layout()
plt.show()
```

```python
# So why does Treatment B look better overall? Let's check the patient distribution.
print("=== HOW PATIENTS WERE DISTRIBUTED ===")
print()
distribution = patients.groupby(["treatment", "stone_size"]).size().unstack()
print(distribution)
print()
print("💡 AHA!")
print("   Treatment A was given to 42 LARGE stone patients (the harder cases)")
print("   Treatment B was given to 42 SMALL stone patients (the easier cases)")
print()
print("   The uneven distribution made Treatment B's overall numbers look better")
print("   — not because it's better, but because it got the easier patients!")
```

### 🤯 Simpson's Paradox — Discussion Points

**Summary table to show students:**

| | Treatment A | Treatment B | Looks better? |
|---|---|---|---|
| **Small stones** | 87.5% | 85.7% | ✅ A |
| **Large stones** | 71.4% | 62.5% | ✅ A |
| **Overall** | 74.0% | 82.0% | ✅ B (?!) |
