## Part 3: Calculations - Computing BMI

Python can do math! Here are the basic operations:
- `+` addition
- `-` subtraction  
- `*` multiplication
- `/` division
- `**` power (e.g., `2 ** 3` means 2³ = 8)

### BMI Formula:
**BMI = weight (kg) ÷ height² (m²)**

In Python: `bmi = weight_kg / (height_m ** 2)`

```python
# BMI Calculation Example
weight_kg = 70.5
height_m = 1.65

# Calculate BMI
bmi = weight_kg / (height_m ** 2)

print("Weight:", weight_kg, "kg")
print("Height:", height_m, "m")
print("BMI:", bmi)
```

```python
# Let's make it look nicer with rounding!
# round(number, decimal_places) rounds to that many decimals

weight_kg = 70.5
height_m = 1.65

bmi = weight_kg / (height_m ** 2)
bmi_rounded = round(bmi, 1)  # Round to 1 decimal place

print("Your BMI is:", bmi_rounded)
```

## Part 5: Making Decisions - Health Categories

Computers can make decisions using `if`, `elif` (else if), and `else`.

**BMI Categories (WHO):**
| BMI | Category |
|-----|----------|
| Below 18.5 | Underweight |
| 18.5 - 24.9 | Normal weight |
| 25.0 - 29.9 | Overweight |
| 30.0 and above | Obese |

**Important:** Notice the spaces (indentation) before `print`. Python needs these!

```python
# Simple example of if/else
bmi = 22.5

if bmi < 18.5:
    print("Underweight")
else:
    print("Not underweight")
```

```python
# Try changing the bmi value and running again!
bmi = 27.3

if bmi < 18.5:
    print("Category: Underweight")
elif bmi < 25:
    print("Category: Normal weight")
elif bmi < 30:
    print("Category: Overweight")
else:
    print("Category: Obese")
```

## 🚀 BONUS CHALLENGE: Patient Triage Tool

If you finish early, try this simple triage tool that checks vital signs!

```python
# 🚨 SIMPLE PATIENT TRIAGE TOOL 🚨

print("╔════════════════════════════════════╗")
print("║    Patient Triage Assessment       ║")
print("╚════════════════════════════════════╝")
print()

# Get vital signs
heart_rate = int(input("Enter heart rate (bpm): "))
temperature = float(input("Enter temperature (°C): "))
pain_level = int(input("Enter pain level (0-10): "))

# Assess urgency
print("\n" + "="*40)
print("TRIAGE ASSESSMENT")
print("="*40)

# Check heart rate
if heart_rate > 120 or heart_rate < 50:
    print("🔴 URGENT: Abnormal heart rate!")
else:
    print("🟢 Heart rate normal")

# Check temperature
if temperature > 39.0:
    print("🔴 URGENT: High fever!")
elif temperature > 37.5:
    print("🟡 MODERATE: Elevated temperature")
else:
    print("🟢 Temperature normal")
    
# Check pain level
if pain_level >= 8:
    print("🔴 URGENT: Severe pain!")
elif pain_level >= 5:
    print("🟡 MODERATE: Significant pain")
else:
    print("🟢 Pain level manageable")

print("="*40)
print("Please review with clinician.")
```
