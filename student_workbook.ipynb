### Loading a Dataset

We've got data from a (fictional) clinical trial comparing two treatments for kidney stones:

- **Treatment A** — open surgery
- **Treatment B** — a newer, less invasive procedure

For each patient, we recorded:
- Which treatment they received
- The size of their kidney stone (`small` or `large`)
- Whether the treatment was successful (`1` = yes, `0` = no)

Let's create this dataset and take a look:

```python
# Clinical trial dataset: 100 patients treated for kidney stones
patients = pd.DataFrame({
    "patient_id": range(1, 101),
    "treatment": ["A"]*50 + ["B"]*50,
    "stone_size": ["small"]*8 + ["large"]*42 + ["small"]*42 + ["large"]*8,
    "success": [1]*7 + [0]*1 + [1]*30 + [0]*12 + [1]*36 + [0]*6 + [1]*5 + [0]*3
})

# .head(10) shows the first 10 rows - like peeking at the top of a spreadsheet
patients.head(10)
```

### Exploring the Data

`.head()` shows the first few rows. A **DataFrame** is just like a spreadsheet — rows and columns.

Let's get an overview of the full dataset:

```python
# How big is our dataset? (rows, columns)
print("Dataset shape:", patients.shape)
print()

# What's the overall success rate?
print("Success rate overall:", round(patients["success"].mean() * 100, 1), "%")
print()

# How many patients got each treatment?
print("Patients per treatment:")
print(patients["treatment"].value_counts())
print()

# How many had small vs large stones?
print("Stone sizes:")
print(patients["stone_size"].value_counts())
```

## Part 4: Filtering Data

One of the most powerful things in pandas is **filtering** — pulling out just the rows you care about.

The syntax is: `dataframe[dataframe["column"] == value]`

Think of it as asking: *"Show me only the rows where this condition is true."*

```python
# Filter: only patients who received Treatment A
treatment_a = patients[patients["treatment"] == "A"]
print("Treatment A patients:")
print(treatment_a.head())
print()

# Filter: only patients with small stones
small_stones = patients[patients["stone_size"] == "small"]
print("Small stone patients:")
print(small_stones.head())
```

## Part 5: The Big Question — Which Treatment is Better?

Let's use `.groupby()` to compare the two treatments. This groups all the rows by a column and lets us calculate statistics for each group — like a pivot table in Excel.

**This is the kind of analysis that gets published in medical journals.**

```python
# Compare overall success rates between treatments
success_by_treatment = patients.groupby("treatment")["success"].mean() * 100
success_by_treatment = round(success_by_treatment, 1)

print("=== SUCCESS RATES BY TREATMENT ===")
print()
print(success_by_treatment.to_string())
print()
print("🤔 So which treatment looks better?")
```

---

## Part 6: Visualisation

Numbers are great, but a chart tells the story instantly. Let's plot the success rates.

`matplotlib` is Python's most popular graphing library.

```python
# Bar chart: overall success rate by treatment
success_by_treatment.plot(kind="bar", color=["#2196F3", "#FF9800"], edgecolor="black")
plt.title("Overall Success Rate by Treatment")
plt.ylabel("Success Rate (%)")
plt.xlabel("Treatment")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print()
print("📊 Looks like Treatment B is the clear winner...")
print("   But should we trust this? Let's dig deeper.")
```

---

## Part 7: The Plot Twist 🔄

What if the overall numbers are **misleading**?

Let's break down the success rates by **both** treatment AND stone size. This is still just one line of pandas:

```python
# Success rates broken down by treatment AND stone size
detailed = patients.groupby(["treatment", "stone_size"])["success"].mean() * 100
detailed = round(detailed, 1)

print("=== SUCCESS RATES: TREATMENT × STONE SIZE ===")
print()
print(detailed.to_string())
print()
print("👀 Wait... look at those numbers carefully.")
print("   Treatment A is better for small stones AND large stones?!")
```

```python
# Let's visualise this properly
detailed_df = detailed.unstack()

detailed_df.plot(kind="bar", color=["#f44336", "#4CAF50"], edgecolor="black")
plt.title("Success Rate by Treatment AND Stone Size")
plt.ylabel("Success Rate (%)")
plt.xlabel("Treatment")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.legend(title="Stone Size")
plt.tight_layout()
plt.show()
```

```python
# So why does Treatment B look better overall? Let's check the patient distribution.
print("=== HOW PATIENTS WERE DISTRIBUTED ===")
print()
distribution = patients.groupby(["treatment", "stone_size"]).size().unstack()
print(distribution)
print()
print("💡 AHA!")
print("   Treatment A was given to 42 LARGE stone patients (the harder cases)")
print("   Treatment B was given to 42 SMALL stone patients (the easier cases)")
print()
print("   The uneven distribution made Treatment B's overall numbers look better")
print("   - not because it's better, but because it got the easier patients!")
```

### 🤯 You Just Discovered Simpson's Paradox!

**Simpson's Paradox** is when a trend that appears in groups of data **reverses** when the groups are combined. You just saw it in action:

| | Treatment A | Treatment B | Looks better? |
|---|---|---|---|
| **Small stones** | 87.5% | 85.7% | ✅ A |
| **Large stones** | 71.4% | 62.5% | ✅ A |
| **Overall** | 74.0% | 82.0% | ✅ B (?!) |

Treatment A is better in **every** subgroup, but worse **overall**. How?

- Treatment A was mostly assigned to patients with **large** stones (harder cases, lower success rates)
- Treatment B was mostly assigned to patients with **small** stones (easier cases, higher success rates)
- So Treatment B's overall average was inflated by having easier patients

**This is based on a real study.** The kidney stone data here is inspired by a [famous 1986 paper by Charig et al.](https://en.wikipedia.org/wiki/Simpson%27s_paradox#Kidney_stone_treatment) that demonstrated exactly this paradox.

### Why This Matters in Medicine

This kind of hidden bias appears everywhere in medical research:
- 💊 **Drug trials** where patient severity isn't evenly distributed
- 🏥 **Hospital league tables** — hospitals that take sicker patients look worse
- 📊 **Treatment comparisons** across different demographics

**The lesson:** Never trust a summary statistic without understanding the data underneath it. Python and pandas give you the tools to look deeper — and you now know how to use them.
