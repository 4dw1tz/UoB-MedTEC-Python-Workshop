## Part 3: Calculations - Computing BMI

Python can do math! Here are the basic operations:
- `+` addition
- `-` subtraction  
- `*` multiplication
- `/` division
- `**` power (e.g., `2 ** 3` means 2³ = 8)

### BMI Formula:
**BMI = weight (kg) ÷ height² (m²)**

In Python: `bmi = weight_kg / (height_m ** 2)`

```python
# BMI Calculation Example
weight_kg = 70.5
height_m = 1.65

# Calculate BMI
bmi = weight_kg / (height_m ** 2)

print("Weight:", weight_kg, "kg")
print("Height:", height_m, "m")
print("BMI:", bmi)
```

```python
# Let's make it look nicer with rounding!
# round(number, decimal_places) rounds to that many decimals

weight_kg = 70.5
height_m = 1.65

bmi = weight_kg / (height_m ** 2)
bmi_rounded = round(bmi, 1)  # Round to 1 decimal place

print("Your BMI is:", bmi_rounded)
```

## Part 5: Making Decisions - Health Categories

Computers can make decisions using `if`, `elif` (else if), and `else`.

**BMI Categories (WHO):**
| BMI | Category |
|-----|----------|
| Below 18.5 | Underweight |
| 18.5 - 24.9 | Normal weight |
| 25.0 - 29.9 | Overweight |
| 30.0 and above | Obese |

**Important:** Notice the spaces (indentation) before `print`. Python needs these!

```python
# Simple example of if/ else if (elif)/ else
bmi = 22.5

if bmi < 18.5:
    print("Underweight")
elif bmi == 18.5:
    print("Exactly 18.5 - just on the edge of underweight!")
else:
    print("Not underweight")
```

## 🎉 Part 6: YOUR Complete BMI Calculator!

Now let's put everything together! Fill in the missing parts marked with `___`

**This is YOUR medical Python program!**

```python
# ========================================
# 🏥 COMPLETE BMI CALCULATOR & ADVISOR 🏥
# ========================================

print("╔════════════════════════════════════╗")
print("║     BMI Calculator & Advisor       ║")
print("║    BAMSoc Conference 2026          ║")
print("╚════════════════════════════════════╝")
print()

# Get patient information
name = input("Enter patient name: ")
weight_kg = float(input("Enter weight in kg: "))
height_m = float(input("Enter height in meters: "))

# Calculate BMI
bmi = weight_kg / (height_m ** 2)
bmi = round(bmi, 1)

# Display results
print("\n" + "="*40)
print("PATIENT REPORT")
print("="*40)
print("Patient:", name)
print("Weight:", weight_kg, "kg")
print("Height:", height_m, "m")
print("BMI:", bmi)
print("-"*40)

# Provide health category and advice
if bmi < 18.5:
    print("📊 Category: UNDERWEIGHT")
    print("💡 Advice: Consider consulting a nutritionist.")
    print("   Ensure adequate caloric intake and nutrition.")
elif bmi < 25:
    print("📊 Category: NORMAL WEIGHT ✅")
    print("💡 Advice: Excellent! Maintain your healthy lifestyle.")
    print("   Continue balanced diet and regular exercise.")
elif bmi < 30:
    print("📊 Category: OVERWEIGHT")
    print("💡 Advice: Consider lifestyle modifications.")
    print("   Increase physical activity, monitor diet.")
else:
    print("📊 Category: OBESE")
    print("💡 Advice: Please consult your healthcare provider.")
    print("   A comprehensive health assessment is recommended.")

print("="*40)
print("Thank you for using BMI Calculator!")
```
